## Task 05: String and list manipulation

### 1. Reverse a sentence word by word
**Logic:**
1. `sentence.split()` turns sentence into a list of words: ex. `"the cat sat"` becomes `["the", "cat", "sat"]`.
2. `words[::-1]` reverses the list: `["sat", "cat", "the"]`.
3. `" ".join(...)` combine the words back into one string with a space between each.

```python
def reverse_words(sentence: str) -> str:
    """return sentence in reverse order."""
    words = sentence.split()
    return " ".join(words[::-1])

print(reverse_words("the cat sat"))
```

### 2. Count how many times each word appears
**Logic:**
1. `counts = {}` creates an empty dictionary.
2. `sentence.lower().split()` makes everything lowercase and splits it into words.
3. The loop visits each word. `counts.get(word, 0)` looks up the word's current count, or gives `0` if the word has not been seen yet.
   dictionary.get(key, default_value)
4. Adding 1 and storing it back updates the count.

```python
def count_words(sentence: str) -> dict:
    """Return a dict of how many times each word appears."""
    counts = {}
    for word in sentence.lower().split():
        counts[word] = counts.get(word, 0) + 1
    return counts

print(count_words("the cat saw the dog"))   
```

### 3. Keep only the even numbers, each doubled
**Logic:** a list comprehension has the shape `[result for item in list if condition]`.
- List comprehension is a shorter way to create a new list using a loop and condition.
- It does the same job as a loop with `.append()`, but in one line.

def doubled_evens(nums):
    result = []

    for x in nums:
        if x % 2 == 0:
            result.append(x * 2)

    return result

print(doubled_evens([1, 2, 3, 4, 5, 6]))

```python
def doubled_evens(nums: list) -> list:
    """Return only the even numbers from nums, each doubled."""
    return [x * 2 for x in nums if x % 2 == 0]

print(doubled_evens([1, 2, 3, 4, 5, 6])) 
```

### 4. Unique items, sorted
**Logic:**
1. `set(items)` removes duplicates, because a set can only hold each value once: `[5, 3, 5, 1, 3, 2]` becomes `{1, 2, 3, 5}`.
2. A set has no order, so `sorted(...)` arranges the values from smallest to largest.

```python
def unique_sorted(items: list) -> list:
    """Return the unique items from the list, sorted."""
    return sorted(set(items))

print(unique_sorted([5, 3, 5, 1, 3, 2])) 
```

### 5. Palindrome check
**Logic:**
1. `word.lower()` makes the word lowercase.
2. `word[::-1]` is the word written from backwards.
3. `word == word[::-1]` compares the two. If they match, the result is `True`.

```python
def is_palindrome(word: str) -> bool:
    """Return True if the word reads the same backwards."""
    word = word.lower()
    return word == word[::-1]

print(is_palindrome("Level"))   
print(is_palindrome("python"))   
```

**What does `word[::-1]` do, and why?**
(Answer: it is a slice operation [start:stop:step]. : start and : stop mean "the whole thing", and a step of -1 means "go backwards".)

**How does a dict make word-counting easier than a list?**
(Answer: a dict stores each word with its key and value. With a list I would have to search through it every time to find the word and its count.)

**When would I reach for a set?**
(Answer: whenever the problem is about uniqueness, like removing duplicates.)

### 1. calculate_stats: min, max, mean and count as a dict
**Logic:**
1. `if not numbers:` checks for an empty list first. An empty list counts as "false" in Python. is empty list then `sum / len` would divide by zero.
2. For an empty list the function returns `{"error": "empty list"}`,instead of a crash.
3. - `sum(numbers) / len(numbers)` is the mean (the total divided by how many numbers there are).
   - `len(numbers)` is the count.
4. The type hint `list[float]` says the input is a list of numbers, and `-> dict` says the output is a dict.

```python
def calculate_stats(numbers: list[float]) -> dict:
    """Return the min, max, mean and count of a list of numbers as a dict.

    If the list is empty, return {'error': 'empty list'} instead of crashing.
    """
    if not numbers:
        return {"error": "empty list"}
    return {
        "min": min(numbers),
        "max": max(numbers),
        "mean": sum(numbers) / len(numbers),
        "count": len(numbers),
    }

print(calculate_stats([88, 92, 79, 95]))   
print(calculate_stats([]))               
print(calculate_stats([7]))              
```

### 2. safe_divide: division that handles zero
**Logic:**
1. The risky operation, `a / b`, goes inside `try:`.
2. If `b` is 0, Python raises a `ZeroDivisionError`. The `except ZeroDivisionError:` block catches it and returns `None` instead of crashing the program.
3. If nothing goes wrong, the `try` block returns the result as normal and the `except` block is skipped.
4. The return type is `float | None`, because the function returns either a number or `None`.

```python
def safe_divide(a: float, b: float) -> float | None:
    """Divide a by b. Return None if b is zero instead of crashing."""
    try:
        return a / b
    except ZeroDivisionError:
        print(f"Cannot divide {a} by zero")
        return None

print(safe_divide(10, 2)) 
print(safe_divide(10, 0))   
```

**Why return a dict rather than four separate values?**
(Answer: a dict keeps key and value pair together. It's also easy to add a fifth value later.)

**What specific error should safe_divide catch, and why not catch everything?**
(Answer: ZeroDivisionError, because that's the one I expect. If I caught everything, real bugs like passing text instead of a number would be hidden and become hard to find.)

**How do type hints and docstrings help the next person?**
(Answer: the type hints show what goes in and what comes out, and the docstring explains what the function does. Someone reading it in six months, including me, can use it without reading the code inside.)
