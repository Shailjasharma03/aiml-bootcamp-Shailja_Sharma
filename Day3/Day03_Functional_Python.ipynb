## Task 01: The ten problems
**What I did:** Solved ten short problems twice each: first with a plain loop, then with a functional tool (comprehension, `map`/`filter`, `reduce`, or `itertools`). For each one I wrote which I prefer and why.

**Logic of the tools used:**
- A list comprehension `[expr for x in items if condition]` builds a new list in one line from a loop plus an `if`.
- `map(func, a, b)` applies a function to items, `filter(func, items)` keeps the ones where the function is true.
- `reduce(func, items)` combines a whole list into one value by applying the function again and again.
- `chain.from_iterable` joins lists end to end, and `accumulate` gives running totals.

```python
from functools import reduce
from itertools import chain, accumulate
from operator import mul
```

```python
nums = [1, 2, 3, 4, 5]

# 1 -- loop version
out = []
for n in nums:
    if n % 2 == 1:
        out.append(n * n)
print(out)

# 1 -- functional version
out = [n * n for n in nums if n % 2 == 1]
print(out)

# Preference: functional, because the whole idea (square, only odd) reads in one line.
```

```python
names = ["  asha ", "Ravi  ", " meera"]

# 2 -- loop version
out = []
for name in names:
    out.append(name.strip().upper())
print(out)

# 2 -- functional version
out = [name.strip().upper() for name in names]
print(out)

# Preference: functional, because the loop only exists to append one transformed item.
```

```python
a = [1, 2, 3]
b = [4, 5, 6]

# 3 -- loop version
out = []
for i in range(len(a)):
    out.append(a[i] * b[i])
print(out)

# 3 -- functional version
out = list(map(mul, a, b))
print(out)

# Preference: functional, because it avoids the index juggling (range(len(a))) of the loop.
```

```python
words = ["cat", "apple", "banana", "dog", "pear", "orange"]

# 4 -- loop version
out = []
for w in words:
    if len(w) > 4:
        out.append(w)
print(out)

# 4 -- functional version
out = list(filter(lambda w: len(w) > 4, words))
print(out)

# Preference: the loop, because filter with a lambda is harder to read than [w for w in words if len(w) > 4] for the same result.
```

```python
nums = [1, 2, 3, 4, 5]

# 5 -- loop version
product = 1
for n in nums:
    product *= n
print(product)

# 5 -- functional version
product = reduce(mul, nums)
print(product)

# Preference: the loop here, because it is obvious at a glance, while reduce needs the reader to know how reduce works.
```

```python
items = [{"name": "pen", "price": 10},
         {"name": "book", "price": 250},
         {"name": "bag", "price": 700}]

# 6 -- loop version
total = 0
for item in items:
    total += item["price"]
print(total)

# 6 -- functional version
total = sum(item["price"] for item in items)
print(total)

# Preference: functional, because it says "sum of the prices" in one clear line.
```

```python
nested = [[1, 2], [3, 4], [5]]

# 7 -- loop version
out = []
for sub in nested:
    for x in sub:
        out.append(x)
print(out)

# 7 -- functional version
out = list(chain.from_iterable(nested))
print(out)

# Preference: functional, because chain.from_iterable names exactly what I want
# (join the sublists) and avoids a nested loop.
```

```python
strings = ["apple", "fig", "banana"]

# 8 -- loop version
out = {}
for s in strings:
    out[s] = len(s)
print(out)

# 8 -- functional version
out = {s: len(s) for s in strings}
print(out)

# Preference: functional (dict comprehension), because it is short and clear.
```

```python
values = [0, 1, "", "a", None, [], [1], False, 2.5]

# 9 -- loop version
out = []
for v in values:
    if v:
        out.append(v)
print(out)

# 9 -- functional version
out = [v for v in values if v]
print(out)

# Preference: functional, because "keep v if v is truthy" is the whole logic.
```

```python
nums = [1, 2, 3, 4]

# 10 -- loop version
out = []
total = 0
for n in nums:
    total += n
    out.append(total)
print(out)

# 10 -- functional version
out = list(accumulate(nums))
print(out)

# Preference: functional, because accumulate does exactly this and needs no extra variable.
```

**Problem 7: is there something in itertools that flattens?**
(Answer: yes, `itertools.chain` (or `chain.from_iterable`). It joins several lists end to end into one stream of items.)

**Problem 10: is there something in itertools for running totals?**
(Answer: yes, `itertools.accumulate`. It returns the running total after each item.)

**Problem 5: why does reduce need an import when map and filter do not?**
(Answer: `map` and `filter` are built-in functions, but `reduce` was moved to the `functools` module in Python 3. It is used less often and can be hard to read, so loops or functions like `sum` are usually clearer.)

## Task 02: Word frequency with Counter
**What I did:** Counted how often each word appears in a text file, first by hand with a dict, then with `Counter`. I then cleaned the text (lowercase, no punctuation) and removed stopwords so the top 10 shows meaningful words.

**Logic:**
- `text.lower().split()` makes a list of lowercase words, but `split()` cuts only on whitespace, so punctuation stays stuck to words (`mistakes.` and `mistakes` look different).
- By hand: a dict where each word is a key and each value is its count. `counts.get(w, 0) + 1` means "the current count, or 0 if I haven't seen it yet, plus one".
- `Counter` does the same counting in one line, and `most_common(n)` returns the top `n` words.
- To clean the text I take only the letters with a regex, and to focus on meaning I drop stopwords (very common words like "the" and "a").

```python
#read the file
with open("text.txt") as f:
    text = f.read().lower()

words = text.split()
print(len(words))
```

```python
#count by hand
counts = {}
for w in words:
    counts[w] = counts.get(w, 0) + 1

print(len(counts))
```

```python
#count by Counter
from collections import Counter

counter = Counter(words)
print(counter == counts)
print(counter.most_common(10))
```

**Comparison:** the by-hand version took 3 lines (create the dict, loop, update the count), and `Counter(words)` took 1 line. Both gave the same result (`counter == counts` is `True`).

**What I notice:** the top words are `the`, `a`, `and`, `to`, `is`. They are common in every text, so they tell me nothing about this one.

```python
#the punctuation problem
print([(w, n) for w, n in counts.items() if w.startswith("mistake")])
```

`mistakes.` (with a full stop) and `mistakes` are counted as different words. `split()` only cuts on spaces, so any punctuation stays attached to the word. The same happens with `the.` and `the`. I need to clean the text first.

```python
#normalise (lowercase and strip punctuation)
import re

clean_words = re.findall(r"[a-z']+", text)
clean_counter = Counter(clean_words)

print(len(clean_words), len(clean_counter))
print(clean_counter.most_common(10))
```

`re.findall(r"[a-z']+", text)` collects runs of letters (and apostrophes) and ignores everything else, so punctuation is gone. The text was already lowercased in Cell 2. The number of different words dropped from 139 to 133 because variants like `mistakes.` and `mistakes` merged, and `that` went from 4 to 5.

```python
#remove stopwords and rerun
stopwords = {"the", "a", "an", "and", "or", "but", "to", "of", "in", "is", "are",
             "it", "that", "they", "who", "will", "for", "them", "what", "so",
             "as", "at", "on", "with", "by", "be", "not", "than", "into",
             "about", "how", "each", "every", "one", "only"}

meaningful = [w for w in clean_words if w not in stopwords]
final_counter = Counter(meaningful)
print(final_counter.most_common(10))
```

Now the top 10 is about the text's content (program, small, words, read, people, learning) instead of filler words. Words with the same count appear in the order they first showed up in the text. A longer text would give a more meaningful top 10.

***Two kinds of “noise”: punctuation (cleaned by the regex) and stopwords (removed by the filter). They are separate problems***

**How do I strip punctuation?**
(Answer: `re.findall(r"[a-z']+", text)` keeps only the letters. Another way is `text.translate(str.maketrans("", "", string.punctuation))`, which removes every punctuation character.)

**What does Counter return for a word that was never seen? Why is that useful?**
(Answer: it returns 0 instead of raising a `KeyError` like a normal dict would. That is useful because I can write `counter[word] += 1` or check a count without first checking whether the word exists.)

**What is a stopword? Where would I get a list of them?**
(Answer: a stopword is a very common word, such as "the", "a" or "and", that carries little meaning for analysis. Ready-made lists exist in libraries like NLTK and scikit-learn, but those need a pip install, so today I wrote a small list myself.)

## Task 03: Combinations with itertools
**What I did:** Used `itertools` to generate all pairs of students with `combinations`, all ordered pairs with `permutations`, every shirt colour and size pairing with `product`, and every subset of 4 dataset features.

**Logic:**
- `combinations(items, r)` picks `r` items where **order does not matter**: (asha, ravi) and (ravi, asha) are the same choice, so only one is kept.
- `permutations(items, r)` picks `r` items where **order matters**, so (asha, ravi) and (ravi, asha) are two different results.
- `product(a, b)` pairs every item of `a` with every item of `b`.
- A subset can have any size from 0 (empty) up to all the items, so for all subsets I loop over every size `r` and collect the `combinations` of that size.

```python
from itertools import combinations, permutations, product

students = ["asha", "ravi", "meera", "dev", "priya"]

pairs = list(combinations(students, 2))
print(len(pairs), "pairs")
print(pairs)
```

**Formula for pairs (nCr):** n! / (r! x (n - r)!)

With n = 5 students and r = 2: 5! / (2! x 3!) = 120 / (2 x 6) = 120 / 12 = **10**

The code gave 10 pairs, so it matches.

```python
from math import factorial

formula = factorial(5) // (factorial(2) * factorial(3))
print(formula, formula == len(pairs))
```

```python
ordered = list(permutations(students, 2))
print(len(ordered), "ordered pairs")
print(ordered[:6])

# Permutations gives 20, combinations gave 10. In permutations the order
# matters, so ('asha', 'ravi') and ('ravi', 'asha') count as two different
# results. In combinations they are the same pair, counted once. Every pair
# shows up twice (once in each order), so the count is exactly double.
```

**Formula for ordered pairs (nPr):** n! / (n - r)!

With n = 5 and r = 2: 5! / 3! = 120 / 6 = **20**. It matches the 20 I got, and it is 2 times the 10 from combinations, because each pair can be ordered in 2! = 2 ways.

```python
colours = ["red", "blue", "green"]
sizes = ["M", "L"]

shirts = list(product(colours, sizes))
print(len(shirts), "shirts")
print(shirts)
```

`product` pairs every colour with every size: 3 colours x 2 sizes = 6 shirts.

```python
features = ["age", "income", "score", "city"]

subsets = []
for r in range(len(features) + 1):      # sizes 0, 1, 2, 3, 4
    for c in combinations(features, r):
        subsets.append(c)

print(len(subsets), "subsets")
for s in subsets:
    print(s)
```

**How it works:** the outer loop tries each subset size `r` from 0 to 4. The inner loop collects all `combinations(features, r)` of that size. Size 0 gives the empty subset `()`, and size 4 gives all four features.

**Why 16?** Each feature is either in a subset or out of it (2 choices), so 4 features give 2 x 2 x 2 x 2 = 2^4 = 16. The counts by size are 1 + 4 + 6 + 4 + 1 = 16. This is what you do in feature selection: try different groups of features and see which works best.

**When does order matter, combinations or permutations? Which gives more results and why?**
(Answer: permutations are for when order matters, like a race ranking, and combinations are for when it doesn't, like choosing a team. Permutations give more results, because every group of items can be arranged in several orders and each order counts separately.)

**What is the formula for nCr? Does my pair count match it?**
(Answer: nCr = n! / (r! x (n - r)!). For 5 students and pairs, that is 10, and my code also gave 10.)

**itertools powerset recipe:**
(Answer: the documented recipe is `chain.from_iterable(combinations(s, r) for r in range(len(s) + 1))`. It does the same thing as my loop over `r`, in one expression.)

## Task 04: Email extraction with regex
**What I did:** Used regular expressions to pull email addresses and phone numbers out of messy text, split each email into username and domain with a capture group, and counted emails per domain with `Counter`.

**Logic:** a regex describes a shape of text. My email pattern is built from three parts:
- `[\w.+-]+` is the username: one or more letters, digits, underscores, dots, plus signs or hyphens.
- `@` is the literal at sign.
- `[\w-]+(?:\.[\w-]+)+` is the domain: a name, followed by one or more `.something` parts, so it needs at least one dot (`gmail.com`, `company.co.in`, `sub.domain.example.org`).

```python
import re
from collections import Counter

text = '''
asha.sharma@codetrade.io wrote to ravi_k99@gmail.com
meera.p+work@company.co.in, phone 022-2555-1234
dev@sub.domain.example.org and +91 98765 43210
not.an.email@ nor @nothing.com -- watch these
'''
```

```python
emails = re.findall(r"[\w.+-]+@[\w-]+(?:\.[\w-]+)+", text)
print(emails)
print(len(emails))
```

**Testing the awkward cases:**
- Dots in the name (`asha.sharma`): `.` is allowed in the username part.
- Plus sign (`meera.p+work`): `+` is allowed in the username part.
- Subdomains (`sub.domain.example.org`): the `(?:\.[\w-]+)+` part repeats for each extra dot.
- Trap 1, `not.an.email@`: it has no domain after the `@`, so the pattern can't complete a match and skips it.
- Trap 2, `@nothing.com`: it has no username before the `@`, so the pattern can't start a match and skips it.

The pattern found exactly the four real emails and skipped both traps.

```python
phones = re.findall(r"(?:\+91[\s-]?)?\d{5}[\s-]?\d{5}|\d{2,4}-\d{3,4}-\d{4}", text)
print(phones)
```

**How the phone pattern works:** the `|` means "or", so it has two formats.
- Format 1: `(?:\+91[\s-]?)?\d{5}[\s-]?\d{5}` is a mobile number, 5 digits and 5 digits, with an optional `+91` in front and an optional space or hyphen between the halves. It matches `+91 98765 43210`.
- Format 2: `\d{2,4}-\d{3,4}-\d{4}` is a landline with an area code, like `022-2555-1234`.

It handles two formats, as the task asks. Other formats (such as `098765-43210`) would need more alternatives.

```python
parts = re.findall(r"([\w.+-]+)@([\w-]+(?:\.[\w-]+)+)", text)

for username, domain in parts:
    print(username, "|", domain)
```

The round brackets `( )` create capture groups: the first one around the username and the second around the domain. When a pattern has groups, `findall` returns a list of tuples with one item per group, so I can unpack each as `username, domain`. The group inside the domain is written `(?:...)`. The `?:` makes it a non-capturing group, so it only groups the pattern without adding an extra item to the tuple.

```python
domain_counts = Counter(domain for username, domain in parts)
print(domain_counts)
```

Every domain appears once, because the sample text has four different domains. With a bigger list of emails the counts would differ, and `most_common()` would show which domain is most frequent.

```python
print(re.findall(r"\bemail", "an email"))
print(re.findall("\bemail", "an email"))
```

```python
print(re.findall(r"[\w.+-]+@[\w-]+(?:\.[\w-]+)+", text))
print(re.search(r"[\w.+-]+@[\w-]+(?:\.[\w-]+)+", text).group())
print(re.match(r"[\w.+-]+@[\w-]+(?:\.[\w-]+)+", text))
```

**Why does every regex pattern start with `r""`? What breaks without it?**
(Answer: `r""` makes a raw string, so Python leaves backslashes alone and passes them to the regex engine unchanged. Without it, Python reads backslash sequences itself. In my test, `"\b"` became a backspace character instead of the "word boundary" regex symbol, so the search for `\bemail` found nothing, while `r"\bemail"` found it.)

**What is the difference between `findall`, `search` and `match`?**
(Answer: `findall` returns a list of all matches in the text. `search` returns only the first match, wherever it is in the text. `match` only looks at the very start of the text. In my test, `match` gave `None` because the text starts with a blank line, and `search` still found the first email.)

**Why should I not validate email with regex? Do I agree?**
(Answer: the email format has many rare but legal cases, so a regex tends to either reject real addresses or accept invalid ones. Even a perfectly shaped address may not exist, and the only real check is to send a confirmation email. I agree: a pattern like mine is fine for pulling emails out of text, but not for deciding whether an address is valid.)

## Task 05: collections bake-off
**What I did:** Solved four small problems, each with the `collections` tool that fits it best: `defaultdict`, `namedtuple`, `deque` and `Counter`.

**Logic:**
- `defaultdict(list)` is a dict that creates an empty list automatically when I use a key it hasn't seen, so I never check for a missing key.
- `namedtuple` is a tuple whose positions have names, so I can write `p.city` instead of `p[2]`.
- `deque(maxlen=5)` is a list-like queue that holds at most 5 items. When a 6th is added, the oldest drops off automatically.
- `Counter` counts items and supports arithmetic: `-`, `&` and `|` compare two counts.

```python
from collections import defaultdict, namedtuple, deque, Counter
```

```python
pairs = [("Sales", "Asha"), ("IT", "Ravi"), ("Sales", "Meera"),
         ("HR", "Dev"), ("IT", "Priya"), ("Sales", "Kiran")]

groups = defaultdict(list)
for dept, emp in pairs:
    groups[dept].append(emp)

print(dict(groups))
```

**How it works:** `defaultdict(list)` gives every new key an empty list as its starting value. So the first time `groups["Sales"]` is used, it is created as `[]` and `.append(emp)` works straight away. With a normal dict this line would raise a `KeyError` unless I checked `if dept not in groups` first.

```python
people = [("Asha", 21, "Ahmedabad"), ("Ravi", 25, "Pune"), ("Meera", 23, "Surat")]

Person = namedtuple("Person", ["name", "age", "city"])
person_list = [Person(*p) for p in people]
print(person_list)

p = person_list[0]
print(p.city)
print(p[2])
```

**How it works:** `namedtuple("Person", ["name", "age", "city"])` creates a new tuple type with named fields. `Person(*p)` unpacks each plain tuple into it.

**Why `p.city` beats `p[2]`:** both give `Ahmedabad`, but `p.city` says what it means, while `p[2]` makes me remember that position 2 is the city. If the order of fields changes, `p[2]` would silently give the wrong thing. The printout also shows the field names, which makes debugging easier.

```python
history = deque(maxlen=5)

for search in ["python", "regex", "pandas", "numpy", "git", "docker"]:
    history.append(search)
    print(list(history))
```

**How it works:** the parameter that makes the dropping automatic is `maxlen`. After 5 searches the deque is full. When the 6th search (`docker`) is added, the oldest (`python`) is pushed out, so the history always holds the last 5. I didn't write any logic to remove anything.

```python
text1 = "the cat sat on the mat and the dog sat too"
text2 = "the dog ran to the park and the cat slept"

c1 = Counter(text1.split())
c2 = Counter(text2.split())

print("c1 - c2:", c1 - c2)
print("c1 & c2:", c1 & c2)
print("c1 | c2:", c1 | c2)
```

```python
common = sorted((c1 & c2).keys())
only_in_text1 = sorted(set(c1) - set(c2))
only_in_text2 = sorted(set(c2) - set(c1))

print("common to both:", common)
print("only in text 1:", only_in_text1)
print("only in text 2:", only_in_text2)
```

**What each operator does:**
- `c1 - c2` subtracts counts and keeps only the words that are left with a count above zero (words that are more frequent in text 1).
- `c1 & c2` is the intersection: words in both, using the smaller count of each.
- `c1 | c2` is the union: all words, using the larger count of each.

For "words that appear in one text but not the other" I use the keys with `set` subtraction, because `c1 - c2` would also keep a word that is in both texts but more frequent in the first.

**What does `maxlen` do to a deque when it is full?**
(Answer: when a new item is added to the right end, the item at the opposite (left) end is dropped automatically, so the deque never grows past `maxlen`.)

**What do `-`, `&` and `|` mean for word counts?**
(Answer: `-` subtracts counts and keeps only positive results, `&` keeps the minimum count of words found in both, and `|` keeps the maximum count across both.)

**When is a namedtuple better than a small class? When worse?**
(Answer: better for simple records that only hold data, because it takes one line, is lightweight, can't be changed by accident, and works like a normal tuple. Worse when I need to change values later, add my own methods or validation, or use inheritance, because a class handles those better.)

## Task 06: Parse a real log file
**What I did:** Generated a log file with 300 lines, then parsed it from raw text with a regex, turned the timestamps into `datetime` objects, and answered questions about it with `Counter` and `defaultdict`.

**Logic of the pipeline:**
1. **Generate:** write 300 lines in the format `YYYY-MM-DD HH:MM:SS LEVEL message`, with the time moving forward and a mix of INFO, WARN and ERROR.
2. **Parse:** a regex with three capture groups pulls out the timestamp, the level and the message from each line.
3. **Convert:** `datetime.strptime` turns the timestamp text into a real `datetime`, so I can read its hour.
4. **Analyse:** `Counter` counts levels, hours and error messages, and `defaultdict(list)` groups the entries by level.


```python
import random
import re
from collections import Counter, defaultdict
from datetime import datetime, timedelta

random.seed(42)    # fixed seed, so I get the same file each time

levels = ["INFO"] * 6 + ["WARN"] * 3 + ["ERROR"] * 2     # INFO most often, ERROR least
messages = {
    "INFO": ["User logged in", "File uploaded", "Report generated", "Session started"],
    "WARN": ["Disk usage above 80%", "Slow response from server", "Retrying failed request"],
    "ERROR": ["Database connection failed", "Payment API timeout",
              "Null value in user record", "Disk full"],
}

current = datetime(2026, 11, 1, 0, 0, 0)
with open("app.log", "w") as f:
    for _ in range(300):
        current += timedelta(seconds=random.randint(1, 600))
        level = random.choice(levels)
        msg = random.choice(messages[level])
        f.write(f"{current:%Y-%m-%d %H:%M:%S} {level} {msg}\n")
```

**How it works:** each line moves the clock forward by a random 1 to 600 seconds, so the entries run in time order over about a day. `random.choice(levels)` picks a level, and the list repeats INFO and WARN so they appear more often than ERROR, like in a real log. A message is then picked from that level's list. `random.seed(42)` makes the random choices repeatable.

```python
pattern = r"(\d{4}-\d{2}-\d{2} \d{2}:\d{2}:\d{2})\s+(\w+)\s+(.*)"

entries = []
with open("app.log") as f:
    for line in f:
        m = re.match(pattern, line)
        if m:
            ts, level, msg = m.groups()
            when = datetime.strptime(ts, "%Y-%m-%d %H:%M:%S")
            entries.append((when, level, msg))

print(len(entries), entries[0])
```

**How the pattern works:**
- `(\d{4}-\d{2}-\d{2} \d{2}:\d{2}:\d{2})` is group 1, the timestamp: 4 digits, 2 digits, 2 digits for the date, then 2:2:2 for the time.
- `\s+(\w+)` is group 2, the level: one or more letters.
- `\s+(.*)` is group 3, the message: everything left on the line.

`m.groups()` returns the three captured pieces. `strptime` then converts the timestamp text into a `datetime`, so each entry is stored as `(datetime, level, message)`. All 300 lines matched.

```python
level_counts = Counter(level for when, level, msg in entries)
print(level_counts)
```

```python
hour_counts = Counter(when.hour for when, level, msg in entries)
hour, count = hour_counts.most_common(1)[0]
print(f"Busiest hour: {hour:02d}:00 ({count} entries)")
```

`when.hour` is the hour (0 to 23) of each entry, which is only possible because I converted the timestamp text into a real `datetime`. `Counter` counts entries per hour, and `most_common(1)` returns the top one. The hours are counted across all dates, so it answers "busiest hour of the day".

```python
error_messages = Counter(msg for when, level, msg in entries if level == "ERROR")
print(error_messages.most_common(1))
```

```python
by_level = defaultdict(list)
for when, level, msg in entries:
    by_level[level].append((when, msg))

print({level: len(items) for level, items in by_level.items()})
print(by_level["ERROR"][0])
```

`defaultdict(list)` creates an empty list for each new level, so `by_level[level].append(...)` works without checking whether the key exists. The group sizes (162 + 49 + 89 = 300) match the `Counter` result, and `by_level["ERROR"][0]` is the first error in the file.

**Why `re.match` here rather than `re.search`? Does it matter for this pattern?**
(Answer: `re.match` only matches at the very start of the line, and every log line should start with its timestamp, so `match` says exactly what I mean. `re.search` would look anywhere in the line. For these clean lines both give the same result, but with `match` a line that only has a timestamp-like text in the middle would be correctly skipped.)

**What does `strptime` need to know about my timestamp format?**
(Answer: it needs a format string describing exactly how the text is laid out, using codes like `%Y` for the year, `%m` for the month, `%d` for the day, `%H:%M:%S` for the time, with the same separators as the text. If the format doesn't match the text, it raises a `ValueError`.)

**For the bonus: subtracting two datetimes gives what type? What can I do with it?**
(Answer: a `timedelta`, which is a length of time. I can compare timedeltas to find the biggest one, convert them with `.total_seconds()`, or add them to a datetime. I haven't done the bonus yet.)
