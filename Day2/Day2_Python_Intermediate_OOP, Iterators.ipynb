## Task 01: The Student class

**Logic:**
- `__init__` runs automatically when I create a student class object. It stores the name and roll number, and starts an empty `marks` list for that student.
- `self` means "this particular student".
- `add_mark` appends a score to this student's list.
- `average`: `if not self.marks` is true for an empty list, so it returns `None` early. Otherwise it divides the total by the count, and `round(..., 2)` keeps two decimals.
- `highest` and `lowest`: `max` and `min` give the biggest and smallest mark. `default=None` is what they return for an empty list, instead of raising an error.

```python
class Student:
    """A student with a name, roll number and a list of marks."""

    def __init__(self, name, roll):  #constructor
        self.name = name             #self refer current object
        self.roll = roll
        self.marks = []        

    def add_mark(self, score):
        """Add one score to this student's marks."""
        self.marks.append(score)

    def average(self):
        """Return the mean of the marks, or None if there are no marks."""
        if not self.marks:
            return None
        return round(sum(self.marks) / len(self.marks), 2)

    def highest(self):
        """Return the highest mark, or None if there are no marks."""
        return max(self.marks, default=None)

    def lowest(self):
        """Return the lowest mark, or None if there are no marks."""
        return min(self.marks, default=None)
```

```python
asha = Student("Asha", 12)   #creating object called asha
ravi = Student("Ravi", 7)
meera = Student("Meera", 21)

for score in [88, 92, 79]:      #adding marks for each student
    asha.add_mark(score)

for score in [70, 65, 80, 75]:
    ravi.add_mark(score)

for score in [95, 85]:
    meera.add_mark(score)

for s in [asha, ravi, meera]:
    print(f"{s.name} (roll {s.roll}): marks={s.marks}, average={s.average()}, "   #printing each student detail and finding highest lowset
          f"highest={s.highest()}, lowest={s.lowest()}")
```

```python
dev = Student("Dev", 30)           # no marks added
print(dev.average(), dev.highest(), dev.lowest())   # None None None
```

**What happens when I write `Student("Asha", 12)`? Which method runs?**
(Answer: Python creates a new empty Student object, then automatically runs `__init__` on it, passing the object as `self` and "Asha" and 12 as `name` and `roll`.)

**What happens if I delete `self` from a method?**
(Answer: I get a `TypeError` saying the method takes 1 positional argument but 2 were given. Python always passes the object as the first argument.)

**Why does `self.marks = []` belong in `__init__` and not at class level?**
(Answer: a list at class level is created once and shared by every student. Inside `__init__`, each new student gets a fresh list of their own.)

## Task 02: Make your object printable
**What I did:** Printed a `Student` object, saw an unreadable result, then added `__str__` and `__repr__` to control how the object appears as text.

**Logic:**
- Python has no idea how I want my own class to look when printed, so by default it shows the class name and a memory address.
- `__str__` is the readable version, meant for people. `print()` and `f"{...}"` use it.
- `__repr__` is the unambiguous version, meant for developers. It usually shows how the object could be created.
- Methods whose names start and end with double underscores are called **dunder methods**. Python calls them automatically in specific situations.

```python
print(asha)
# Output (ugly): <__main__.Student object at 0x000001A2B3C4D5E0>
# It only shows the class name and a memory address, which tells me nothing about the student.
```

```python
class Student:
    """A student with a name, roll number and a list of marks."""

    def __init__(self, name, roll):
        self.name = name
        self.roll = roll
        self.marks = []

    def add_mark(self, score):
        """Add one score to this student's marks."""
        self.marks.append(score)

    def average(self):
        """Return the mean of the marks, or None if there are no marks."""
        if not self.marks:
            return None
        return round(sum(self.marks) / len(self.marks), 2)

    def highest(self):
        """Return the highest mark, or None if there are no marks."""
        return max(self.marks, default=None)

    def lowest(self):
        """Return the lowest mark, or None if there are no marks."""
        return min(self.marks, default=None)

    def __str__(self):
        """Readable text, for people."""
        return f"{self.name} (roll {self.roll})"

    def __repr__(self):
        """Precise text, for developers."""
        return f"Student(name={self.name!r}, roll={self.roll!r})"


# the old objects still use the old class, so recreating the students
asha = Student("Asha", 12)
ravi = Student("Ravi", 7)
meera = Student("Meera", 21)
```

**How the two methods work:**
- `__str__` returns a friendly string like `Asha (roll 12)`.
- `__repr__` returns `Student(name='Asha', roll=12)`, which looks like the code that would build the object. The `!r` inside the f-string adds quotes around text, so a name shows as `'Asha'` and a number stays `12`. !r tells python to use repr() representation.


```python
print(asha)           # uses __str__
print(str(asha))      # same: __str__
print(repr(asha))     # uses __repr__
```

```python
students = [asha, ravi, meera]
print(students)
```

**Which one runs inside a list: `__str__` or `__repr__`?**

`__repr__`. The list printed `Student(name='Asha', roll=12)`, not `Asha (roll 12)`.

**Why?** When Python prints a list, it shows each item using `repr()`. A list is a container, and it uses the precise, unambiguous form so you can tell what's inside (for example, `'5'` the text is different from `5` the number).

**What is the difference between `__str__` and `__repr__`? Who is each for?**

(Answer: `__str__` is the readable version for end users, used by `print()`. `__repr__` is the precise version for developers,store inside containers like lists.)

**Why does printing a list not use `__str__` on the items?**
(Answer: a list's own `__str__` is built to call `repr()` on each item, so the contents stay unambiguous.)

**Two other dunder methods I could add to Student:**

## Task 03: Inheritance, and knowing when not to
**What I did:** Created a `Person` class, then made `Student` and `Teacher` inherit from it. `Student` overrides `greet()`, and I wrote down whether inheritance is the right choice here.

**Logic:**
- `Person` is the parent class. It holds what every person has: a name and a `greet()` method.
- `class Student(Person):` means Student gets everything Person has, and can add more.
- `super().__init__(name)` runs the parent's `__init__`, so the name is stored by the code that already knows how to do it. Then the child adds its own data (`roll`).
- `Teacher` also inherits from `Person` but has its own extra data, `subject`.
- `Overriding` means a child class defines a method with the same name with same parameter as the parent's. Python then uses the child's version for that class.

```python
class Person:
    """A person with a name."""

    def __init__(self, name):
        self.name = name

    def greet(self):
        """Return a general greeting."""
        return f"Hello, I am {self.name}."
```

```python
class Student(Person):
    """A person with a roll number."""

    def __init__(self, name, roll):
        super().__init__(name)
        self.roll = roll

    def greet(self):
        """Override Person.greet with a student greeting."""
        return f"Hi, I'm {self.name}, roll number {self.roll}. I'm a student!"
```

```python
class Teacher(Person):
    """A person who teaches a subject."""

    def __init__(self, name, subject):
        super().__init__(name)
        self.subject = subject
```

**How it works:**
- `Student.__init__` calls `super().__init__(name)` first. That runs `Person.__init__`, which sets `self.name`. Then the Student sets its own `roll` and `marks`.
- `Student.greet` has the same name as `Person.greet`, so it overrides it. A Student gives its own greeting.

```python
person = Person("Ramesh")
student = Student("Asha", 12)
teacher = Teacher("Mrs. Shah", "Mathematics")

print(person.greet())
print(student.greet())
print(teacher.greet())
print(teacher.subject)
```

```python
# Should Student inherit from Person, or just have a person?
#
# My position: Student should inherit from Person.
#
# A student really is a person, so inheritance fits naturally here. It avoids
# repeating name and greet(), and the hierarchy is only one level deep, so it is
# Composition (self.person = Person(name)) would add extra code
# for no real benefit in a small program like this. I would switch to
# composition if roles started to change or combine (for example, someone who is
# both a student and a teacher), because deep inheritance gets hard to change.
```

**What does `super()` actually do? What breaks if I forget to call it?**
(Answer: `super()` gives access to the parent class, so `super().__init__(name)` runs the parent's setup. If I forget it, the parent's attributes like `self.name` are never created, and using them raises an `AttributeError`.)

**Why do experienced developers say "composition over inheritance"?**
(Answer: inheritance ties a child tightly to its parent, so a change in the parent can break every child, and deep hierarchies are hard to follow. Composition builds an object from smaller parts, which is more flexible and easier to change.)

**What is method overriding? What happens to the parent's version?**
(Answer: overriding is when a child class defines a method with the same name as the parent's. Python uses the child's version for child objects. The parent's version still exists and can be reached with `super().method()`.)

## Task 04: Take a for loop apart
**What I did:** Got an iterator from a list with `iter()`, pulled values from it with `next()` until it ran out, then rebuilt what a `for` loop does using only `iter`, `next`, `try/except` and `break`.

**Logic:**
- `iter([1, 2, 3])` gives an iterator, an object that remembers its position in the list.
- Each `next(it)` returns the next item and moves the position forward.
- When there are no items left, `next(it)` raises an error. A `for` loop quietly catches that error and stops, so I never see it.
- My `while True` loop does by hand what `for` does automatically: get an iterator, call `next` repeatedly, and stop when the error appears.

```python
it = iter([1, 2, 3])
print(next(it))
print(next(it))
print(next(it))
```

```python
try:
    print(next(it))      # a fourth call: nothing left
except Exception as e:
    print("Error name:", type(e).__name__)
```

```python
for x in [1, 2, 3]:
    print(x)
```

```python
it = iter([1, 2, 3])
while True:
    try:
        x = next(it)
    except StopIteration:
        break
    print(x)
```

**How the `while True` loop reproduces `for`:**
1. `it = iter([1, 2, 3])` gets the iterator, like the `for` loop does at the start.
2. `while True:` repeats forever, until something breaks out.
3. `x = next(it)` takes the next value. If there is none, it raises `StopIteration`.
4. `except StopIteration: break` catches that error and ends the loop. This is the step a `for` loop does silently.
5. `print(x)` runs only when a value was found, so the output is the same as the `for` loop: 1, 2, 3.

**Result:** both loops print the same output, and the error is called `StopIteration`.S

```python
#for look it up question i created this code block
import io

f = io.StringIO("line one\nline two\n")   # behaves like an open text file
print(iter(f) is f)       # True means it is its own iterator
print(next(f))            
print(next(f))            
```

**What exception does a finished iterator raise? Why does a normal for loop never show it?**
(Answer: `StopIteration`. A `for` loop catches it internally and treats it as the signal to stop, so it never reaches my code.)

**What is the difference between an iterable and an iterator?**
(Answer: an iterable is something I can loop over, like a list, string or dict. It gives me an iterator when I call `iter()` on it. An iterator is the object that actually produces the values one at a time with `next()`, and it remembers where it is. Once an iterator is used up, it's finished.)

**Is a file object an iterator? Test it.**
(Answer: yes. `iter(f) is f` was `True`, and `next(f)` returned one line at a time. That is why `for line in f` reads a file line by line.)

## Task 05: Generators and yield
**What I did:** Wrote the Fibonacci sequence two ways, as a function that returns a list and as a generator that uses `yield`, then compared them.

**Logic:**
- Fibonacci starts 0, 1, and each next number is the sum of the previous two. `a, b = b, a + b` moves both values forward in one step.
- `fib_list(n)` builds the whole list in memory and returns it.
- `fib_gen(n)` uses `yield`. It hands back one number at a time, pauses, and continues from the same spot when the next number is asked for. It never builds the full list.

```python
def fib_list(n):
    result = []
    a, b = 0, 1
    for _ in range(n):
        result.append(a)
        a, b = b, a + b
    return result
```

```python
def fib_gen(n):
    a, b = 0, 1
    for _ in range(n):
        yield a
        a, b = b, a + b
```

```python
print(fib_list(10))
print(list(fib_gen(10)))
print(fib_list(10) == list(fib_gen(10)))
```

```python
print(type(fib_list(10)))
print(type(fib_gen(10)))
```

`fib_list(10)` returns a **list**: all 10 numbers already exist in memory. `fib_gen(10)` returns a **generator object**: no numbers exist yet, and each one is created only when it is asked for.

```python
gen = fib_gen(5)

for x in gen:
    print(x)

print("second loop:")
for x in gen:
    print(x)

# The second loop prints nothing because a generator can only be used once.
# It remembers where it stopped, and after the first loop it has reached
# the end, so there is nothing left to give. To loop again, I must create
# a new generator with fib_gen(5).
```

```python
def fib_gen(n):
    a, b = 0, 1
    for _ in range(n):
        print("about to yield", a)   
        yield a
        a, b = b, a + b
```

```python
gen = fib_gen(3)
print("generator created")

for x in gen:
    print("got", x)
```

"generator created" printed first, with no "about to yield" before it. So calling `fib_gen(3)` does not run any of the function's code. The code runs only when the loop asks for a value: it runs up to the `yield`, hands the value over, then pauses. On the next request it continues right after the `yield`. That is why "about to yield" and "got" alternate.

```python
import sys

print(sys.getsizeof(fib_list(1000)))
print(sys.getsizeof(fib_gen(1000)))
```

**What does `yield` do that `return` does not?**
(Answer: `return` sends back a value and ends the function for good. `yield` sends back a value and pauses the function, keeping its local variables, so it can carry on from the same place the next time a value is requested.)

**Why does looping over a generator twice give nothing the second time?**
(Answer: a generator is an iterator, so it is used up after one pass. It remembers only its current position, and once it reaches the end it has no values left. A new generator must be created to start again.)

**Generator memory:**
(Answer: `sys.getsizeof` showed the list is much bigger than the generator, and its size grows with n. The generator stays small because it creates one value at a time instead of storing them all. That is useful for big data.)

## Task 06: Read a big file without loading it
**What I did:** Generated a text file with 500,000 lines, then counted its lines and the lines containing one word, reading one line at a time and never calling `.read()`.

**Logic:**
- `with open(...) as f:` opens the file and closes it automatically when the block ends, so I never call `.close()` myself.
- A file object is an iterator (Task 04), so `for line in f` gives me one line at a time. Only one line is in memory at once, however big the file is.
- `f.read()` would load the entire file as one huge string, and `.split()` would then build a second huge list. That's what causes memory trouble.

```python
topics = ["python", "data", "code", "file", "loop"]

with open("big.txt", "w") as f:
    for i in range(1, 500001):
        f.write(f"line {i} talks about {topics[i % 5]}\n")
```

```python
with open("big.txt") as f:
    count = 0
    for line in f:
        count += 1

print(count)
```

```python
with open("big.txt") as f:
    matches = 0
    for line in f:
        if "python" in line:
            matches += 1

print(matches)
```

**How it works:**
- The loop visits each line once. For the line count, `count += 1` runs every time. For the word count, it runs only when `"python" in line` is true.
- Every 5th line ends with `python` (`topics[i % 5]` cycles through the five words), so 500,000 / 5 = 100,000 matching lines. That matches the output, which confirms the code is right.
- Neither cell calls `.read()` or `.close()`.

**What does the `with` statement actually guarantee? What is a context manager?**
(Answer: `with` guarantees the file is closed when the block ends, even if an error happens inside it. `f.closed` printed `True` after the block. A context manager is an object that sets something up when the block starts and cleans up when it ends, which is how `open()` works.)

**What is the difference between `read()`, `readline()` and `readlines()`?**
(Answer: `read()` returns the whole file as one string. `readline()` returns one line each time it is called. `readlines()` returns a list of all the lines. `read()` and `readlines()` load everything into memory, while looping with `for line in f` reads one line at a time.)

**Why should I never parse CSV by splitting on commas? Find a value that breaks it.**

**Why is the line-by-line loop better than `f.read().split()`?**
(Answer: it keeps only one line in memory at a time, so it works for files of any size. `f.read().split()` loads the whole file and then builds a second big list, which can use too much memory.)
